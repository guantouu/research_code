```python
import torch
import math
import matplotlib.pyplot as plt
import csv
```

```python
input_file = open('/app/layer_record_VGG8/inputConv6_.csv')

input_row_1 = list(csv.reader(input_file))[0]
input_item = {}

for x in input_row_1:
    y = str(x).strip()
    if y not in input_item.keys():
        input_item[y] = 1
    else:
        input_item[y] += 1

index_i = list(range(len(input_item)))
plt.bar(index_i, input_item.values())
plt.show()
print(index_i)
print(input_item.keys())
```

```python
weight_file = open('/app/layer_record_VGG8/weightConv6_.csv')

weight_row = list(csv.reader(weight_file))
weight_item = {}

for row in weight_row:
    for x in row:
        y = str(x).strip()
        if y not in weight_item.keys():
            weight_item[y] = 1
        else:
            weight_item[y] += 1

index_i = list(range(len(weight_item)))
plt.bar(index_i, weight_item.values())
plt.show()
print(len(index_i))
print(weight_item.keys())
```
